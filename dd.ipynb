```python
df['City'] = df['City'].replace('BBSR','Bhubaneshwar')
df['City'] = df['City'].replace('Bhubaneswar','Bhubaneshwar')
df['City'].unique()
```

```python
df['Floor_No'].unique()
```

```python
df['Floor_No'] = df['Floor_No'].replace({'Ground':'0','G':'0'})
df['Floor_No'].unique()
df['Floor_No'] = df['Floor_No'].astype(int)

df['Floor_No'].dtypes 

```

```python
# df['Garage'].unique()
df['Garage'] = df['Garage'].replace({'YES':'1','No':'0','Y':'1','N':'0','Yes':'1'})
df['Garage'].unique()
df['Garage'] = df['Garage'].astype(int)
df['Garage'] = df['Garage'].astype(bool)
df['Garage']
```

```python
df['Furnished'] = df['Furnished'].replace({'YES':'1','No':'0','Y':'1','N':'0','Yes':'1'})
df['Furnished'].unique()
df['Furnished'] = df['Furnished'].astype(int)
df['Furnished'] = df['Furnished'].astype(bool)
df['Furnished']
```

```python
df['Property_Type'] = df['Property_Type'].replace({'APT':'Apartment'})
df['Property_Type'].unique()
```
