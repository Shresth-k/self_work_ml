```python
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

```

```python
#load the dataset
data = load_diabetes()
data
df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"]=data.target
df
```

```python


X = df.drop("target",axis=1)
y_with_N = df.target
```

```python
X
```

```python
y_with_N
```

```python
y_binary = (y_with_N > 120).astype(int)
y_binary
y_binary.value_counts()
```

```python
#split into training and testing data
Xtrain,Xtest,ytrain,ytest = train_test_split(X,y_binary,test_size=0.3,random_state=42)
```

```python
#model training
model = GaussianNB()
model.fit(Xtrain,ytrain)
```

```python
ypred = model.predict(Xtest)
ypred
```

```python
print("Training score: ",model.score(Xtrain,ytrain))
print("Testing score: ",model.score(Xtest,ytest))
```

```python
con_matrix = confusion_matrix(ytest,ypred)
con_matrix
```

```python

print("Accuracy: ",accuracy_score(ytest,ypred))
print("Precision: ",precision_score(ytest,ypred))
print("Recall: ",recall_score(ytest,ypred))
print("F1: ",f1_score(ytest,ypred))
```
